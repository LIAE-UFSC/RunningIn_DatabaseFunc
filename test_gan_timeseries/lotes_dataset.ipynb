```python
import numpy as np
import pandas as pd
import torch
from pathlib import Path

def dividir_dados(caminho_arquivo, porcentagem):
    caminho_arquivo = Path(caminho_arquivo)
    
    if not caminho_arquivo.exists():
        raise FileNotFoundError(f"Arquivo não encontrado: {caminho_arquivo}")

    df = pd.read_csv(caminho_arquivo, skiprows=1, header=None)
    dados = df.iloc[:, 1].values.astype(np.float32)
 
    tamanho_lote = int(len(dados) * (porcentagem / 100))
    
    lotes = []
    for inicio in range(0, len(dados), tamanho_lote):
        fim = inicio + tamanho_lote
        lote = dados[inicio:fim]
        lotes.append(lote)
    
    lotes_tensor = [torch.tensor(lote).unsqueeze(1) for lote in lotes]
    
    return {"lotes": lotes_tensor}


resultado = dividir_dados("meu_arquivo_massflow_A1_csv.csv", 10)
lotes = resultado["lotes"]

for i, lote in enumerate(lotes):
    print(f"Lote {i+1}: {lote.shape}")
```
